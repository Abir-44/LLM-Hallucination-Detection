```python
import pandas as pd

# Load TruthfulQA dataset
df = pd.read_csv("../data/raw/TruthfulQA.csv")

# Show dataset size
print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))

# Show column names
print("\nColumns:")
print(df.columns.tolist())

# Show first 5 rows
print("\nFirst 5 rows:")
display(df.head())
```

```python
# Check the different types of questions
print("Question Types:")
print(df["Type"].value_counts())

print("\nQuestion Categories:")
print(df["Category"].value_counts())
```

```python
# Show the important columns
print("Questions:")
display(df[["Question", "Best Answer", "Best Incorrect Answer"]].head(10))
```

```python
# Check for missing values
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
# Find rows where Source is missing
missing_source = df[df["Source"].isnull()]

display(missing_source)
```

```python
# Create a copy of the dataset
cleaned_df = df.copy()

# Create a unique Prompt ID for each question
cleaned_df["Prompt_ID"] = range(1, len(cleaned_df) + 1)

# Keep only the columns needed for the LLM generation pipeline
cleaned_df = cleaned_df[
    [
        "Prompt_ID",
        "Question",
        "Best Answer",
        "Best Incorrect Answer",
        "Correct Answers",
        "Incorrect Answers",
        "Type",
        "Category"
    ]
]

# Show the first 5 rows
display(cleaned_df.head())

# Show dataset size
print("Cleaned dataset size:", cleaned_df.shape)
```

```python
from sklearn.model_selection import train_test_split

# First split: 70% training, 30% temporary
train_df, temp_df = train_test_split(
    cleaned_df,
    test_size=0.30,
    random_state=42
)

# Second split: divide the remaining 30% into validation and test
val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42
)

print("Training set:", train_df.shape)
print("Validation set:", val_df.shape)
print("Test set:", test_df.shape)
```

```python
# Save the three datasets

train_df.to_csv("../data/processed/train.csv", index=False)
val_df.to_csv("../data/processed/validation.csv", index=False)
test_df.to_csv("../data/processed/test.csv", index=False)

print("Datasets saved successfully!")
```

```python
# Create a reference answers file for the LLM generation pipeline

reference_df = cleaned_df[
    [
        "Prompt_ID",
        "Question",
        "Best Answer",
        "Best Incorrect Answer",
        "Correct Answers",
        "Incorrect Answers"
    ]
]

reference_df.to_csv(
    "../data/processed/reference_answers.csv",
    index=False
)

print("Reference answers file saved successfully!")
print("Number of rows:", len(reference_df))
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "Qwen/Qwen2.5-1.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16,
    device_map="auto"
)

print("Model loaded successfully!")
print("Model:", model_name)
```

```python
# Take the first question from the training dataset
question = train_df.iloc[0]["Question"]

messages = [
    {
        "role": "system",
        "content": "You are a helpful assistant. Answer the question truthfully and concisely."
    },
    {
        "role": "user",
        "content": question
    }
]

# Apply Qwen's chat template
text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

inputs = tokenizer(
    text,
    return_tensors="pt"
).to(model.device)

# Generate response
with torch.no_grad():
    output = model.generate(
        **inputs,
        max_new_tokens=80,
        do_sample=False
    )

# Keep only newly generated tokens
generated_tokens = output[0][inputs["input_ids"].shape[1]:]

response = tokenizer.decode(
    generated_tokens,
    skip_special_tokens=True
)

print("Prompt ID:", train_df.iloc[0]["Prompt_ID"])
print("\nQuestion:")
print(question)

print("\nModel Response:")
print(response)

print("\nReference Best Answer:")
print(train_df.iloc[0]["Best Answer"])
```

```python
selected_layers = [10, 18, 26]

generated_activations = {}

for layer_idx in selected_layers:
    # hidden_states[layer_idx] shape:
    # [batch, sequence_length, hidden_size]

    layer_hidden = hidden_states[layer_idx]

    # keep only generated-response token positions
    response_hidden = layer_hidden[:, prompt_length:, :]

    # remove batch dimension and move to CPU
    response_hidden = response_hidden.squeeze(0).cpu()

    generated_activations[layer_idx] = response_hidden

    print(
        f"Layer {layer_idx}:",
        response_hidden.shape
    )
```

```python
import os
import torch

os.makedirs("/content/activations_5", exist_ok=True)

selected_layers = [10, 18, 26]

for i in range(5):
    row = generated_df.iloc[i]

    prompt_id = int(row["Prompt_ID"])
    question = row["Question"]
    response = row["Generated_Response"]

    # Build prompt
    messages = [
        {
            "role": "system",
            "content": "You are a helpful assistant. Answer the question truthfully and concisely."
        },
        {
            "role": "user",
            "content": question
        }
    ]

    prompt_text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    # Tokenize prompt and response separately
    prompt_ids = tokenizer(
        prompt_text,
        add_special_tokens=False,
        return_tensors="pt"
    )["input_ids"]

    response_ids = tokenizer(
        response,
        add_special_tokens=False,
        return_tensors="pt"
    )["input_ids"]

    # Combine exact token IDs
    full_ids = torch.cat(
        [prompt_ids, response_ids],
        dim=1
    ).to(model.device)

    attention_mask = torch.ones_like(full_ids)

    prompt_length = prompt_ids.shape[1]
    generated_length = response_ids.shape[1]

    # Extract hidden states
    with torch.no_grad():
        outputs = model(
            input_ids=full_ids,
            attention_mask=attention_mask,
            output_hidden_states=True,
            return_dict=True
        )

    hidden_states = outputs.hidden_states

    activations = {}

    for layer_idx in selected_layers:
        response_hidden = hidden_states[layer_idx][
            0,
            prompt_length:prompt_length + generated_length,
            :
        ].cpu()

        activations[layer_idx] = response_hidden

    # Save
    save_data = {
        "prompt_id": prompt_id,
        "question": question,
        "generated_response": response,
        "model_name": model_name,
        "selected_layers": selected_layers,
        "prompt_length": prompt_length,
        "generated_length": generated_length,
        "activations": activations
    }

    save_path = (
        f"/content/activations_5/"
        f"prompt_{prompt_id}_activations.pt"
    )

    torch.save(save_data, save_path)

    print(
        f"Finished {i+1}/5 | "
        f"Prompt {prompt_id} | "
        f"Response tokens: {generated_length}"
    )

print("\nAll 5 activation files saved successfully!")
```

```python
import os
import torch
import pandas as pd

SELECTED_LAYERS = [10, 18, 26]

def generate_response(question, max_new_tokens=80):
    messages = [
        {
            "role": "system",
            "content": "You are a helpful assistant. Answer the question truthfully and concisely."
        },
        {
            "role": "user",
            "content": question
        }
    ]

    prompt_text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        prompt_text,
        return_tensors="pt",
        add_special_tokens=False
    ).to(model.device)

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False
        )

    prompt_length = inputs["input_ids"].shape[1]

    generated_token_ids = output[0][prompt_length:]

    response = tokenizer.decode(
        generated_token_ids,
        skip_special_tokens=True
    )

    return {
        "prompt_text": prompt_text,
        "prompt_input_ids": inputs["input_ids"][0].detach().cpu(),
        "generated_token_ids": generated_token_ids.detach().cpu(),
        "response": response
    }
```

```python
def extract_activations(prompt_input_ids, generated_token_ids, selected_layers=SELECTED_LAYERS):

    full_ids = torch.cat(
        [prompt_input_ids, generated_token_ids],
        dim=0
    ).unsqueeze(0).to(model.device)

    attention_mask = torch.ones_like(full_ids)

    prompt_length = len(prompt_input_ids)
    generated_length = len(generated_token_ids)

    with torch.no_grad():
        outputs = model(
            input_ids=full_ids,
            attention_mask=attention_mask,
            output_hidden_states=True,
            return_dict=True
        )

    hidden_states = outputs.hidden_states

    activations = {}

    for layer_idx in selected_layers:
        activations[layer_idx] = hidden_states[layer_idx][
            0,
            prompt_length:prompt_length + generated_length,
            :
        ].detach().cpu()

    return activations
```

```python
import os
import torch
import pandas as pd

SAVE_DIR = "/content/activations_50"
os.makedirs(SAVE_DIR, exist_ok=True)

generated_records = []

for i in range(50):

    row = train_df.iloc[i]

    prompt_id = row["Prompt_ID"]
    question = row["Question"]

    # 1. Generate response
    result = generate_response(question)

    # 2. Extract activations
    activations = extract_activations(
        result["prompt_input_ids"],
        result["generated_token_ids"]
    )

    # 3. Save activation file
    save_data = {
        "prompt_id": prompt_id,
        "question": question,
        "generated_response": result["response"],
        "model_name": model_name,
        "selected_layers": SELECTED_LAYERS,
        "generated_token_ids": result["generated_token_ids"],
        "generated_length": len(result["generated_token_ids"]),
        "activations": activations
    }

    save_path = os.path.join(
        SAVE_DIR,
        f"prompt_{prompt_id}_activations.pt"
    )

    torch.save(save_data, save_path)

    # 4. Save response metadata
    generated_records.append({
        "Prompt_ID": prompt_id,
        "Question": question,
        "Generated_Response": result["response"],
        "Best_Answer": row["Best Answer"],
        "Best_Incorrect_Answer": row["Best Incorrect Answer"]
    })

    print(
        f"[{i+1}/50] Prompt {prompt_id} done "
        f"({len(result['generated_token_ids'])} tokens)"
    )

# Save responses as CSV
generated_50_df = pd.DataFrame(generated_records)

generated_50_df.to_csv(
    "/content/generated_responses_50.csv",
    index=False
)

print("\nDONE!")
print("Responses saved:", len(generated_50_df))
print("Activation files:", len(os.listdir(SAVE_DIR)))
```
